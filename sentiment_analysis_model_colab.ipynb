# Sentiment Analysis Model — Google Colab

This notebook loads the provided `sentiment_analysis_model.pkl` and uses it for sentiment prediction.

**Saved model:** scikit-learn Pipeline with TF-IDF word/character features and a LinearSVC classifier.

```python
import joblib

model_path = "sentiment_analysis_model.pkl"
model = joblib.load(model_path)
            
print("Model loaded successfully!")
print(model)
```

```python
# Check the sentiment labels/classes
try:
    print("Classes:", model.named_steps["svm"].classes_)
except Exception:
    print("Could not read classes directly from the model.")
```

```python
# Predict sentiment for sample texts
texts = [
    "I absolutely loved this product!",
    "This was a terrible experience.",
    "The service was okay, nothing special."
]

predictions = model.predict(texts)
            
for text, prediction in zip(texts, predictions):
    print(f"Text: {text}")
    print(f"Sentiment: {prediction}")
    print()
```

```python
# Interactive prediction
text = input("Enter a review/sentence: ")
prediction = model.predict([text])[0]
            
print("Predicted sentiment:", prediction)
```
